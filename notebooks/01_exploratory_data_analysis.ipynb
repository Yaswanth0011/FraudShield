# 🔍 SentinelShield AI — Exploratory Data Analysis (EDA)

This notebook explores the **European Cardholders Credit Card Fraud Detection Dataset** (Worldline & ULB ML Group).

### Objectives:
1. Inspect class distribution and quantify imbalance.
2. Analyze transaction amount and temporal patterns.
3. Investigate PCA latent features and their correlation with fraudulent activity.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style
sns.set_theme(style="whitegrid", palette="muted")
%matplotlib inline
```

```python
# Load dataset
df = pd.read_csv("../data/fraud_data_simulated.csv")
print("Dataset Shape:", df.shape)
df.head()
```

```python
# Class Distribution
class_counts = df['Class'].value_counts()
print(class_counts)
print(f"Fraud Percentage: {class_counts[1] / len(df) * 100:.2f}%")

plt.figure(figsize=(6, 4))
sns.countplot(x='Class', data=df, palette=["#2ecc71", "#e74c3c"])
plt.title("Transaction Class Distribution (0 = Legit, 1 = Fraud)")
plt.show()
```

```python
# Transaction Amount Distribution by Class
fig, axes = plt.subplots(1, 2, figsize=(14, 4))
sns.histplot(df[df['Class'] == 0]['Amount'], bins=50, ax=axes[0], color='#2ecc71', log_scale=True)
axes[0].set_title("Legitimate Transactions - Amount (Log Scale)")

sns.histplot(df[df['Class'] == 1]['Amount'], bins=50, ax=axes[1], color='#e74c3c', log_scale=True)
axes[1].set_title("Fraudulent Transactions - Amount (Log Scale)")
plt.show()
```

```python
# Key PCA Component Distributions (V14, V17, V12, V10)
key_features = ['V14', 'V17', 'V12', 'V10']
fig, axes = plt.subplots(1, 4, figsize=(18, 4))
for i, col in enumerate(key_features):
    sns.boxplot(x='Class', y=col, data=df, ax=axes[i], palette=["#2ecc71", "#e74c3c"])
    axes[i].set_title(f"{col} vs Class")
plt.tight_layout()
plt.show()
```

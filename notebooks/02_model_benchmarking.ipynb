# 📈 FraudShield — Model Benchmarking & Evaluation

This notebook trains and compares baseline supervised learning algorithms (Logistic Regression, Random Forest, XGBoost) on the Credit Card Fraud dataset.

### Disclaimers & Attribution:
- Dataset: Worldline and the Machine Learning Group (ULB).
- Classifiers: Standard implementations from `scikit-learn` and `xgboost`.

```python
import sys
sys.path.append('..')

import pandas as pd
import numpy as np
from src.data_preprocessing import preprocess_and_split
from src.model_training import train_baseline_models
from src.evaluation import evaluate_model_performance
```

```python
# Preprocess and split
X_train, X_test, y_train, y_test, scaler = preprocess_and_split(
    filepath="../data/fraud_data_simulated.csv",
    test_size=0.2,
    random_state=42
)
print(f"Train shape: {X_train.shape}, Test shape: {X_test.shape}")
```

```python
# Train baselines
models = train_baseline_models(X_train, y_train)

# Evaluate and compare
results = []
for name, model in models.items():
    m = evaluate_model_performance(model, X_test, y_test)
    results.append({
        "Model": name,
        "Accuracy": m["accuracy"],
        "Precision": m["precision"],
        "Recall": m["recall"],
        "F1-Score": m["f1_score"],
        "ROC-AUC": m["roc_auc"],
        "PR-AUC": m["pr_auc"]
    })

comparison_df = pd.DataFrame(results)
comparison_df.sort_values(by="PR-AUC", ascending=False)
```
